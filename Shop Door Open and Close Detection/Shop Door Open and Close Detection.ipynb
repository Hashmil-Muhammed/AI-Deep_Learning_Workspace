# **Shop Door Open/Close Detection Using Python and OpenCV**

## **1. Project objective**

The system should perform these operations:

**Daytime monitoring**
- Monitor the shop door.
- Detect when the door opens.
- Display DOOR OPEN.
- Save the opening date and time in a log.
- Generate an opening alert.

**Night-time monitoring**
- Monitor the shop door.
- Detect when the door closes.
- Display DOOR CLOSED.
- Save the closing date and time in a log.
- Generate a closing alert.
- Generate a warning if the door remains open during configured closing hours.

 ## **2. Technologies used**

| Technology | Purpose |

## **4. Important concept – Door Masking (ROI)**

ROI means Region of Interest.

Instead of processing the complete CCTV video, we select only the portion containing the shop door.


The system will:
1. Allow you to select the door area using your mouse.
2. Crop that region from every video frame.
3. Compare the cropped image with known OPEN and CLOSED door images.
4. Identify the current door state.

We will use two reference images:
- open_reference.jpg
- closed_reference.jpg
These are captured directly from your uploaded video during calibration.

## **5. Code Working**


This implementation has an interactive calibration process:
1. Select the door area.
2. Pause the video when the door is OPEN and press O.
3. Pause the video when the door is CLOSED and press C.
4. Press SPACE to resume video playback.
5. OpenCV compares the current door image against both reference images.
6. It detects state changes, generates alerts and saves logs.

### **OpenCL (Open Computing Language)** 

OpenCL (Open Computing Language) is ``a framework that allows programs to use GPUs and CPUs to perform calculations in parallel, making processing faster.``

In simple words: ```OpenCL helps us use the GPU to process many image pixels simultaneously instead of processing them one by one on the CPU.```

### **AMD GPU (Advanced Micro Devices Graphics Processing Unit)**

AMD GPU (Advanced Micro Devices Graphics Processing Unit) is a ````graphics processor designed by AMD that can perform many calculations simultaneously.````

In your project: The AMD Radeon R5 M430 GPU uses OpenCL to process door image comparisons in parallel, helping accelerate image processing.
